```python
import numpy as np
import pandas as pd

# 1. Generate Data
np.random.seed(11)
n = 288  # 24 hours, one reading every 5 minutes
times = pd.date_range("2026-10-13 00:00", periods=n, freq="5min")
hours = times.hour + times.minute / 60
truth_temp = (
    20 + 2.5 * np.sin((hours - 9) / 24 * 2 * np.pi) + np.random.normal(0, 0.1, n)
)
truth = pd.DataFrame({"timestamp": times, "temp_c": truth_temp})

sensor = truth.copy()
drop = np.random.choice(sensor.index, size=34, replace=False)  # ~12% lost packets
sensor = sensor.drop(index=drop).reset_index(drop=True)

stuck_start = 150  # sensor freezes for 14 readings
sensor.loc[stuck_start : stuck_start + 13, "temp_c"] = sensor.loc[
    stuck_start, "temp_c"
]
safe = [i for i in sensor.index if not (stuck_start <= i <= stuck_start + 13)]
g = np.random.choice(safe, size=3, replace=False)  # 3 glitches
sensor.loc[g, "temp_c"] = [-40.0, 85.0, 0.0]

# --- INSPECTION ---
print(f"Total expected readings (n): {n}")
print(f"Readings that arrived (sensor length): {len(sensor)}")
print(f"Missing readings (dropped packets): {n - len(sensor)}")

# Impossible range check (Valid indoor range: 5°C to 40°C)
impossible_mask = (sensor["temp_c"] < 5.0) | (sensor["temp_c"] > 40.0)
num_impossible = impossible_mask.sum()
print(f"Impossible readings detected: {num_impossible}")

# --- STAGE 1: EDGE STAGE (Cheap Checks Only) ---
# Removes out-of-range values locally at the sensor/gateway
edge_cleaned = sensor.copy()
edge_cleaned.loc[
    (edge_cleaned["temp_c"] < 5.0) | (edge_cleaned["temp_c"] > 40.0), "temp_c"
] = np.nan

# --- STAGE 2: CLOUD STAGE (Heavier Steps & History) ---
# Requires historical buffer / rolling window analysis
cloud_cleaned = edge_cleaned.copy()

# Detect stuck sensor: rolling std dev of 0 across 6 consecutive readings
rolling_std = cloud_cleaned["temp_c"].rolling(window=6, min_periods=6).std()
stuck_indices = cloud_cleaned.index[rolling_std == 0]

# Mark stuck readings as NaN
cloud_cleaned.loc[stuck_indices, "temp_c"] = np.nan

# Re-index to full timestamp grid and interpolate missing gaps
full_grid = pd.DataFrame({"timestamp": times})
cloud_cleaned = pd.merge(
    full_grid, cloud_cleaned, on="timestamp", how="left"
)
cloud_cleaned["temp_c"] = cloud_cleaned["temp_c"].interpolate(
    method="linear"
)
```
